```python
#invalid value in product
products[
    products["margin_usd"] !=
    (products["price_usd"] - products["cost_usd"])
]
```

```python
#check in orders col
print(orders["payment_method"].value_counts(dropna=False))
print(orders["device"].value_counts(dropna=False))
print(orders["source"].value_counts(dropna=False))
```

```python
#invalid in order_items
print(order_items[order_items["quantity"] < 0])
print(order_items[order_items["quantity"] == 0])
print(order_items[order_items["unit_price_usd"] < 0])
print(order_items[order_items["line_total_usd"] < 0])
```
